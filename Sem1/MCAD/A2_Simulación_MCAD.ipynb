## Ejercicio 1

Diseñar un experimento de simulación de Montecarlo para aproximar el área bajo la curva de la función presentada en la imagen adjunta.

ENTREGABLES: PDF con código depurado (depurado y subido en forma individual)  

$$ f(x) = x^2 - 5x + 7$$

CurvaSM.jpg

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
n = 10000

x = np.random.uniform(0, 5, n)
y = np.random.uniform(0, 7, n)

fx = x**2 - 5*x + 7

debajo = y <= fx

area = 35 * np.sum(debajo) / n

print(f"Selected area: {area} u^2")
print(f"Selected area (pct): {area * 100/ 35}%")
```

```python
# plot points
plt.scatter(x, y)
plt.scatter(x[debajo], y[debajo])
# add function as curve
x_curve = np.linspace(0, 5, 100)
y_curve = x_curve**2 - 5*x_curve + 7
plt.plot(x_curve, y_curve, color="red")
```

## Ejercicio 2

Diseñar un experimento de simulación de Montecarlo para aproximar el área entre las curvas representadas por las funciones mostradas en la imagen adjunta (2CurvasSM.jpg)  

$$g(x) = - x + 7$$

2CurvasSM.jpg

```python
n = 10000

x = np.random.uniform(0, 5, n)
y = np.random.uniform(0, 7, n)

fx = x**2 - 5*x + 7
gx = -x + 7

debajo = (y >= fx) & (y <= gx)

area = 35 * np.sum(debajo) / n

print(f"Selected area: {area} u^2")
print(f"Selected area (pct): {area * 100/ 35}%")
```

```python
# plot points
plt.scatter(x, y)
plt.scatter(x[debajo], y[debajo])
# add function as curve
x_curve = np.linspace(0, 5, 100)
y_curve = x_curve**2 - 5*x_curve + 7
g_curve = -x_curve + 7
plt.plot(x_curve, g_curve, color="red")
plt.plot(x_curve, y_curve, color="red")
```
